## My Lane: Content Refresh Prioritization

I'm choosing the content refresh prioritization lane. In Week 1, I found a negative
correlation between `avg_position` and `ctr` — pages ranking lower tend to get fewer
clicks, and the CTR drop was sharpest between position tiers 1-3 and 4-10. This suggests
that some pages sitting just outside the top positions may be strong refresh candidates:
a small ranking improvement could recover a disproportionate amount of CTR. I want to dig
deeper into which pages are the best candidates for refresh effort, rather than treating
all "underperforming" pages the same.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## Research Question

**Question:** Among pages that are underperforming for their position, which
characteristics (content_type, content_age_days, trend_direction) best predict whether
refreshing that page is likely to be worth the effort?

**Unit of analysis:** a single page.

**Decision it informs:** which pages a content team should prioritize for a refresh
(rewrite, update, re-optimize) out of a large backlog.

**Action someone takes:** a content editor pulls the top-ranked pages from the refresh
queue and schedules them for updates this sprint, instead of refreshing pages at random
or by gut feeling.

**Cost of a wrong recommendation:** if the model flags the wrong pages, the team spends
limited editorial hours refreshing content that won't move CTR or rankings, while pages
that would have benefited from a refresh keep losing traffic. Wasted effort has a real
opportunity cost since refresh capacity is limited.

**Why data/ML helps:** with tens of thousands of pages, no one can manually judge refresh
priority page-by-page. A model that scores pages by predicted CTR recovery potential turns
a subjective backlog into a ranked, defensible priority list.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd


df = pd.read_csv("/content/content_refresh_anonymized.csv")


corr = df['avg_position'].corr(df['ctr'])
print(f"Correlation between avg_position and ctr: {corr:.3f}")


near_miss = df[(df['avg_position'] > 10) & (df['avg_position'] <= 20)]
print(f"Pages ranking 11-20 (near-miss zone): {len(near_miss)} out of {len(df)} total ({len(near_miss)/len(df)*100:.1f}%)")


top10_ctr = df[df['avg_position'] <= 10]['ctr'].mean()
near_miss_ctr = near_miss['ctr'].mean()
print(f"Avg CTR for top 10: {top10_ctr:.3f} | Avg CTR for positions 11-20: {near_miss_ctr:.3f}")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

## What I Can and Can't Claim

**Observed:** There is a measurable negative correlation between `avg_position` and
`ctr`, and pages in the 11-20 position range show a noticeably lower average CTR than
top-10 pages. A meaningful share of pages sit in this "near-miss" zone.

**Cannot claim:** I cannot say refreshing these pages will *cause* a ranking or CTR
improvement — this is an observed, directional pattern in historical data, not a proven
causal effect. Other factors (content quality, competition, seasonality) could explain
part of the gap. Any model built on this will produce a *prioritization signal*, not a
guarantee, and should be validated against real refresh outcomes before being trusted at
scale. No client-identifying data is used or published.
